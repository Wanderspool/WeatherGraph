# Interactive Visualization with WeatherGraph

This notebook demonstrates how to use the `weathergraph.vis` module to create interactive Leaflet maps and MP4/GIF animations from a forecast dataset.

```python
import xarray as xr
import numpy as np
from weathergraph.vis import create_interactive_map, create_animation

# Create a dummy forecast dataset for demonstration
ds = xr.Dataset(
    {'t': (['time', 'lat', 'lon'], np.random.rand(5, 180, 360))},
    coords={
        'time': np.arange(5), 
        'lat': np.linspace(-90, 90, 180), 
        'lon': np.linspace(-180, 180, 360)
    }
)
```

## 1. Interactive Leaflet Maps

We can use `create_interactive_map` to generate a `folium.Map` object. In a Jupyter environment, this object will automatically render as an interactive map.

```python
# Render the interactive map for the first time step (time_index=0)
interactive_map = create_interactive_map(ds, variable='t', time_index=0, cmap_name='viridis')

# Display the map (you will see it if you run this in Jupyter)
interactive_map
```

## 2. Animations (MP4 / GIF)

We can use `create_animation` to generate an MP4 or GIF across the time dimension of our dataset. This is useful for analyzing the progression of weather patterns.

```python
from IPython.display import Video

# Generate an MP4 animation
output_video_path = 'forecast_animation.mp4'
create_animation(ds, variable='t', output_path=output_video_path, format='mp4', fps=2)

# Display the generated video
Video(output_video_path, embed=True)
```
